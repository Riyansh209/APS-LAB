```python
import matplotlib.pyplot as plt
```

```python
def add_two_numbers(a, b):
    return a + b
n_values = list(range(1, 101))
time_values = []

for n in n_values:
    add_two_numbers(n, n + 1)
    time_values.append(1)

plt.figure()
plt.plot(n_values, time_values)
plt.xlabel("Input size n")
plt.ylabel("Time")
plt.title("O(1) Constant Time Complexity for Addition")
plt.show()
```

```python
def linear_time_sum(numbers):
    total = 0
    for value in numbers:
        total += value
    return total
n_values = list(range(1, 101))
time_values = []
for n in n_values:
    data = list(range(n))
    linear_time_sum(data)
    time_values.append(n)
plt.figure()
plt.plot(n_values, time_values)
plt.xlabel("Input size n")
plt.ylabel("Time")
plt.title("O(n) Linear Time Complexity")
plt.show()
```

```python
import time

def count_positive(numbers):
    count = 0
    for value in numbers:
        if value > 0:
            count += 1
    return count
n_values = list(range(1, 5000, 500))
time_values = []
for n in n_values:
    data = list(range(-n//2, n//2))
    start = time.time()
    count_positive(data)
    end = time.time()
    time_values.append(end - start)
plt.figure()
plt.plot(n_values, time_values)
plt.xlabel("Input size n")
plt.ylabel("Execution Time (seconds)")
plt.title("O(n) Linear Time Complexity (Count > 0)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import math
def binary_search(arr, target):

    n = len(arr)
    if n == 0:
        return 0
    return math.log2(n) if n > 1 else 1

n_values = list(range(1, 101))
time_values_log = []

for n in n_values:
    data = list(range(n))
    time_values_log.append(binary_search(data, 0))
plt.figure(figsize=(10, 6))
plt.plot(n_values, time_values_log, label='O(log n)')
plt.xlabel("Input size n")
plt.ylabel("Time (logarithmic scale)")
plt.title("O(log n) Logarithmic Time Complexity (Binary Search)")
plt.legend()
plt.grid(True)
plt.show()
```
