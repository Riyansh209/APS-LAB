# Travelling Salesman Problem using Dynamic Programming.
Problem Description:
Given a set of cities and distances, find the shortest possible route that visits each city once and returns to the starting city.

#Theory
TSP uses dynamic programming with bit masking. Store results of visited subsets to reduce repeated work.

## Algorithm
1. Start from source city
2. Mark visited cities using bit mask
3. Try all unvisited cities
4. Choose minimum cost path
5. Return to source city

```python
def tsp(graph):
    n = len(graph)
    dp = [[-1]*(1<<n) for _ in range(n)]

    def visit(city, mask):
        if mask == (1<<n) - 1:
            return graph[city][0]

        if dp[city][mask] != -1:
            return dp[city][mask]

        ans = float('inf')

        for next_city in range(n):
            if mask & (1<<next_city) == 0:
                new_cost = graph[city][next_city] + visit(next_city, mask | (1<<next_city))
                if new_cost < ans:
                    ans = new_cost

        dp[city][mask] = ans
        return ans

    return visit(0, 1)
```

```python
graph = [
[0, 10, 15, 20],
[10, 0, 35, 25],
[15, 35, 0, 30],
[20, 25, 30, 0]
]

print("Minimum travelling cost:", tsp(graph))
```

```python
def tsp_with_path(graph):
    n = len(graph)

    dp = [[-1]*(1<<n) for _ in range(n)]

    parent = [[-1]*(1<<n) for _ in range(n)]

    def visit(city, mask):
        if mask == (1<<n) - 1:

            return graph[city][0]

        if dp[city][mask] != -1:
            return dp[city][mask]

        ans = float('inf')
        best_next_city = -1


        for next_city in range(n):

            if not (mask & (1<<next_city)):
                new_cost = graph[city][next_city] + visit(next_city, mask | (1<<next_city))
                if new_cost < ans:
                    ans = new_cost
                    best_next_city = next_city

        dp[city][mask] = ans
        parent[city][mask] = best_next_city
        return ans

    min_cost = visit(0, 1)

    path = []
    current_city = 0
    current_mask = 1
    while True:
        path.append(current_city)
        if current_mask == (1<<n) - 1:

            path.append(0)
            break

        next_city = parent[current_city][current_mask]
        if next_city == -1:
            break
        current_city = next_city
        current_mask |= (1<<next_city)

    return min_cost, path
```

```python
graph_example = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

min_cost, path = tsp_with_path(graph_example)
print("Minimum travelling cost:", min_cost)
print("Path of cities visited:", path)
```

```python
def parse_graph_input(graph_str):
    graph = []
    try:
        rows = graph_str.strip().split(';')
        for row_str in rows:
            row = [int(x.strip()) for x in row_str.split(',')]
            graph.append(row)
        return graph
    except ValueError:
        print("Error: Invalid input format. Please ensure all values are integers and correctly formatted.")
        return None

user_graph = parse_graph_input(user_input_graph_str)

if user_graph:
    print("Parsed Graph:", user_graph)
else:
    print("Could not parse the graph. Please re-enter in the correct format.")
```

```python
if user_graph:
    min_cost_user, path_user = tsp_with_path(user_graph)
    print("\nMinimum travelling cost (User Input):", min_cost_user)
    print("Path of cities visited (User Input):", path_user)
else:
    print("Cannot calculate TSP as the graph was not parsed successfully.")
```
